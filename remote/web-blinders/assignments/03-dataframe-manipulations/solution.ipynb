# Assignment 03 Solution

## Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Create Dataset

```python
df = pd.DataFrame(np.random.randn(10, 4),
                  index=pd.date_range('2021-07-18', periods=10),
                  columns=['a', 'b', 'c', 'd'])
display(df)
```

## Task 1

```python
# Task 1: Create a new column and store the logarithmic values for column d
# Constraint: Without using np.log() directly on Series

def log_value(x):
    return x if x <= 0 else np.log(x)

df['log_d'] = [log_value(x) for x in df['d']]
display(df)
```

## Task 2

```python
# Task 2: Create a new column where:
# Negative values of column d remain same
# Positive values are replaced with their log values

df['transformed_d'] = [x if x < 0 else np.log(x) for x in df['d']]
display(df)
```

## Task 3

```python
# Task 3: Create a column from column a that returns:
# 'yes' if the value is even (after converting to integer)
# 'no' otherwise

df['even_a'] = ['yes' if int(x) % 2 == 0 else 'no' for x in df['a']]
display(df)
```

## Task 4

```python
# Task 4: Create a new column with index values

df['index_col'] = [i for i in df.index]
display(df)
```

## Task 5

```python
# Task 5: Split index (date) into three new columns: year, month, and day

df['year'] = [str(i).split('-')[0] for i in df.index]
df['month'] = [str(i).split('-')[1] for i in df.index]
df['day'] = [str(i).split('-')[2].split(' ')[0] for i in df.index]
display(df)
```

## Task 6

```python
# Task 6: Create a column that returns the weekday name of the index manually (without using .day_name())

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df['weekday'] = [days[i.weekday()] for i in df.index]
display(df)
```
