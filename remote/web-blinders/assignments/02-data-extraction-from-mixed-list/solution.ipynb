# Assignment 02 Solution

## Given Data

```python
l = [
    [1, 2, 3, 4, 5, 6],
    (1, 2, 3, 4, 5, 6),
    (3, 4, 5, 66, 78, 5),
    {
        "k1": "Hello",
        "k2": "how",
        "k3": "are",
        "k4": "you"
    }
]
```

## Task 1

Extract all elements from the list object.

```python
list_elements = l[0]

print(list_elements)
```

## Task 2

Extract the dictionary from the mixed list.

```python
dict_elements = l[3]

print(dict_elements)
```

## Task 3

Extract all tuple objects.

```python
tuple_elements = [item for item in l if isinstance(item, tuple)]

print(tuple_elements)
```

## Task 4

Extract all numerical values from the list and tuples.

```python
numerical_data = []

for item in l:
    if isinstance(item, (list, tuple)):
        numerical_data.extend(item)

print(numerical_data)
```

## Task 5

Calculate the sum of all numerical values.

```python
total_sum = sum(numerical_data)

print(total_sum)
```

## Task 6

Filter and display all odd numbers.

```python
odd_values = [n for n in numerical_data if n % 2 != 0]

print(odd_values)
```

## Final Output

```python
print("1. List Elements:", list_elements)
print("2. Dict Elements:", dict_elements)
print("3. Tuple Elements:", tuple_elements)
print("4. Numerical Data:", numerical_data)
print("5. Sum of Numerical Data:", total_sum)
print("6. Odd Values:", odd_values)
```
